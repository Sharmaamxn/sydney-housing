# Sydney Housing Price Prediction and Decision Support System

This notebook demonstrates the data loading, preprocessing, model tuning, and evaluation workflow for the Sydney housing dataset.

```python
import pandas as pd
from pathlib import Path

data_path = Path('..') / 'data' / 'raw' / '8D_Housing_Data_Template (1).csv'
df = pd.read_csv(data_path)
print(df.shape)
print(df['suburb'].value_counts().to_string())
print(df.head().to_string(index=False))
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold, cross_val_score
from src.housing_model import load_and_prepare_data, make_features, build_model_pipeline

df = load_and_prepare_data()
X, y = make_features(df)
model = build_model_pipeline(LinearRegression())
cv = KFold(n_splits=5, shuffle=True, random_state=42)
scores = -cross_val_score(model, X, y, cv=cv, scoring='neg_root_mean_squared_error')
print(scores)
print(f'Average RMSE: {scores.mean():,.0f}')
```
